## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load starter CSV dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Total dataset count and declining percentage
total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
declining_pct = (declining_pages / total_pages) * 100

# 2. Median staleness (days since update) for declining vs healthy pages
median_stale_down = df[df["trend_direction"].str.lower() == "down"]["days_since_last_update"].median()
median_stale_other = df[df["trend_direction"].str.lower() != "down"]["days_since_last_update"].median()

# 3. High-exposure declining pages (>= 500 impressions)
high_vis_decay = df[(df["impressions_90d"] >= 500) & (df["trend_direction"].str.lower() == "down")].shape[0]

print(f"Total Pages in Dataset: {total_pages:,}")
print(f"Metric 1: {declining_pages:,} pages ({declining_pct:.1f}%) are currently in a declining trend.")
print(f"Metric 2: Declining pages have a median stale age of {median_stale_down:.0f} days vs {median_stale_other:.0f} days for non-declining pages.")
print(f"Metric 3: There are {high_vis_decay:,} high-visibility pages (>=500 impressions) actively declining that represent immediate refresh opportunities.")
```
